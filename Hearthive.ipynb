## **Importing libraries**

```python
from langchain_community.document_loaders import PyPDFDirectoryLoader
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain_community.embeddings import SentenceTransformerEmbeddings
from langchain.vectorstores import Chroma
from langchain_community.llms import LlamaCpp
from langchain.chains import RetrievalQA, LLMChain
```

## **Import the document**


```python
loader=PyPDFDirectoryLoader("/content/drive/MyDrive/mypro/data")
docs=loader.load()
```

```python
len(docs) # number of pages
```

```python
docs[6]
```

## **Chunking**

```python
text_splitter=RecursiveCharacterTextSplitter(chunk_size=300,chunk_overlap=50)
chunks=text_splitter.split_documents(docs)
```

```python
len(chunks)
```

```python
chunks[600]
```

```python
chunks[601]
```

## **Embeddings creations**

```python
import os
os.environ['HUGGINGFACEHUB_API_TOKEN']="YOUR_HF_TOKEN_HERE"
```

```python
embeddings = SentenceTransformerEmbeddings(model_name="NeuML/pubmedbert-base-embeddings")
```

## **Vector Store creation**

```python
vectorstore = Chroma.from_documents(chunks,embeddings)
```

```python
query = "who is at risk of heart disease?"

search_results = vectorstore.similarity_search(query)
```

```python
search_results
```

```python
retriever = vectorstore.as_retriever(search_kwargs={'k':5})
```

```python
retriever.get_relevant_documents(query)
```

## **LLM Model loading**

```python
llm=LlamaCpp(
    model_path="/content/drive/MyDrive/mypro/BioMistral-7B.Q4_K_M.gguf",
    temperature=0.2,
    max_tokens= 2048,
    top_p=1

)
```

## **Use LLM and retriever and query , to genearte final response**

```python
template = """
<|context|>
You are an Medical Assistant that follows the instructions and generate the accurate response based on the query and the context provided.
Please be truthful and give direct answers.
</s>
<|user|>
{query}
</s>
<|assistant|>
"""
```

```python
from langchain.schema.runnable import RunnablePassthrough
from langchain.schema.output_parser import StrOutputParser
from langchain.prompts import ChatPromptTemplate
```

```python
prompt = ChatPromptTemplate.from_template(template)
```

```python
rag_chain = (
  {"context":retriever, "query": RunnablePassthrough()}
  | prompt
  | llm
  | StrOutputParser()
)
```

```python
response = rag_chain.invoke(query)
```

```python
response
```
